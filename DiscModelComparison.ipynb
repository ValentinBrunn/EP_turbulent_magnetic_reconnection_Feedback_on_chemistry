```python
import numpy as np
import matplotlib.pyplot as plt

mH = 1.6735575e-24  # g

def dominant_ion_map(model, nH, m_n=2.2*mH, exclude=("e-",), min_x=0.0):
    """
    Dominant ion map based on ion number density n_i = x_i * nH (not conductivity).
    Returns:
      dom_idx : (nx,nz) index of dominant ion in ion_names
      dom_n   : (nx,nz) dominant ion number density n_i [cm^-3]
      dom_x   : (nx,nz) dominant ion abundance x_i
      ion_names : list of ion names
    """
    mass_dict = model.species.mass
    charge_dict = model.species.charge

    ion_names, ion_mass, ion_Z = [], [], []
    for name, m_i in mass_dict.items():
        name = str(name).strip()
        if name in exclude:
            continue
        Z_i = int(charge_dict.get(name, 0))
        if Z_i == 0:
            continue
        ion_names.append(name)
        ion_mass.append(float(m_i))
        ion_Z.append(Z_i)

    nx, nz = nH.shape
    dom_idx = np.full((nx, nz), -1, dtype=int)
    dom_n = np.zeros((nx, nz), dtype=float)   # stores max n_i
    dom_x = np.zeros((nx, nz), dtype=float)   # stores x_i of the winner

    for j, name in enumerate(ion_names):
        try:
            x_i = model.getAbun(name)  # (nx,nz)
        except Exception:
            continue

        if min_x > 0:
            x_i = np.where(x_i >= min_x, x_i, 0.0)

        n_i = x_i * nH   # cm^-3

        mask = n_i > dom_n
        dom_n[mask] = n_i[mask]
        dom_idx[mask] = j
        dom_x[mask] = x_i[mask]

    return dom_idx, dom_n, dom_x, ion_names


def plot_dominant_ion_names(dom_idx, dom_field, ion_names,
                            step=4, min_val=0.0,
                            title="Dominant ion"):
    """
    dom_field : map used for thresholding text display (e.g. dom_n or dom_x)
    """
    nx, nz = dom_idx.shape

    plt.figure(figsize=(10, 6))
    plt.imshow(dom_idx.T, origin="lower", aspect="auto", cmap="tab20")
    plt.xlabel("ix")
    plt.ylabel("iz")
    plt.title(title)
    plt.colorbar(label="Ion index")

    for ix in range(0, nx, step):
        for iz in range(0, nz, step):
            j = dom_idx[ix, iz]
            if j < 0:
                continue
            if dom_field[ix, iz] <= min_val:
                continue
            plt.text(ix, iz, ion_names[j],
                     ha="center", va="center",
                     fontsize=6, color="black")

    plt.tight_layout()
    plt.show()


# ---- Run ----
dom_idx, dom_n, dom_x, ion_names = dominant_ion_map(
    model,
    model.nHtot,
    exclude=("e-",),
    min_x=1e-30
)

# plot with threshold on dominant number density
plot_dominant_ion_names(
    dom_idx,
    dom_n,
    ion_names,
    step=4,
    min_val=0.0,
    title="Dominant ion by number density"
)
```

```python
import numpy as np

# cgs
ecgs = 4.803204712570263e-10  # statC
mH   = 1.6735575e-24          # g

def sigma_grain_distribution(
    nH,
    Tgas,
    amin,
    amax,
    Z_of_a,
    delta=0.01,
    rho_s=2.0,      # g/cm3 (densité matériau grain)
    m_n=2.2*mH,
    Na=200
):
    """
    Conductivité grains intégrée sur une distribution MRN n(a)=C*(a/a0)^-3.5.
    amin, amax, a0 en cm.
    Z_of_a : fonction qui prend un array a (cm) et renvoie Z(a) (adimensionnel).
    nH [cm^-3], Tgas [K].
    Retour: sigma_g (nx,nz)
    """

    # grille en taille (log-spaced)
    a = np.logspace(np.log10(amin), np.log10(amax), Na)  # cm

    # Normalisation MRN par la masse de poussière: rho_d = delta * rho_n
    rho_n = m_n * nH  # g cm^-3

    # C tel que ∫ m_g(a) n(a) da = delta * rho_n
    # ∫ a^-0.5 da = 2( sqrt(amax) - sqrt(amin) )
    denom = (8.0*np.pi/3.0) * rho_s * (np.sqrt(amax) - np.sqrt(amin))  # g cm^-2.5
    C = (delta * rho_n) / denom  # unités: cm^-0.5 (cohérent avec n(a)~C a^-3.5)

    # Distribution en nombre: n(a) = C * (a/a0)^-3.5
    # -> n(a) a en cm => n(a) en cm^-4 (car n(a) da => cm^-3)
    n_a = C[..., None] * (a[None, None, :] )**(-3.5)

    # Propriétés grains
    m_g = (4.0/3.0)*np.pi*(a**3)*rho_s  # g, shape (Na,)

    # collision rate parameter (ton sigmav)
    # sigmav = 3e-3 * a^2 * sqrt(T/100)
    sigmav = 3e-3 * (a[None, None, :]**2/(1e-12)) * np.sqrt(Tgas[..., None]/100.0)

    gamma = sigmav / (m_n + m_g[None, None, :])  # (nx,nz,Na)

    # Charge des grains
    Z = Z_of_a(a)  # shape (Na,) ou broadcastable
    Z = np.asarray(Z)[None, None, :]  # (1,1,Na)

    # Intégrande: (e^2/rho_n) * Z^2 * n(a) / (m_g * gamma)
    integrand = (ecgs**2 / rho_n[..., None]) * (Z**2) * n_a / (m_g[None, None, :] * gamma)

    # Intégration sur a
    sigma_g = np.trapezoid(integrand, a, axis=-1)  # (nx,nz)

    return sigma_g


Zconst = lambda a: np.ones_like(a)*10  # Z(a)=100

sigma_g = sigma_grain_distribution(
    nH=densiteHmoyliste,
    Tgas=TemperatureGas,
    amin=0.05*1e-4,     # 0.05 micron = 5e-6 cm
    amax=10000*1e-4,   # 10 micron = 1e-3 cm
    Z_of_a=Zconst,
    delta=0.01,
    rho_s=2.076,     # g/cm3 (densité matériau grain Prodimo)
    m_n=2.2*mH,
    Na=300
)
```

```python
import numpy as np

# cgs constants
e_cgs = 4.803204712570263e-10   # statC
c_cgs = 2.99792458e10           # cm/s
pi = np.pi
mH = 1.6735575e-24              # g



def Ambipolar_conductivities_ions(model, nH, B, m_n=2.2*mH,
                        exclude=("e-","PAH","PAH-","PAH+","PAH++","PAH+++")):
    """
    Computes sigma_O (ions-only), sigma_P (Pedersen), sigma_H (Hall),
    sigma_perp, eta_O, eta_A for ions-only.

    Parameters
    ----------
    nH : (nx,nz) neutral hydrogen density proxy (cm^-3)
    B  : (nx,nz) magnetic field strength in Gauss
    """
    # densities
    rho_n = nH * m_n  # g cm^-3

    mass_dict = model.species.mass
    charge_dict = model.species.charge

    # sums for Pedersen/Hall
    sumP = np.zeros_like(nH, dtype=float)
    sumH = np.zeros_like(nH, dtype=float)

    # optional: ion mass density for D (if you want); else D=1
    rho_i = np.zeros_like(nH, dtype=float)

    # safety: avoid division by zero if B has zeros
    Bsafe = np.where(B > 0, B, np.nan)

    for name, m_j in mass_dict.items():
        name = str(name).strip()
        if name in exclude:
            continue

        Z_j = abs(int(charge_dict.get(name, 0)))
        if Z_j == 0:
            continue

        try:
            x_j = model.getAbun(name)  # (nx,nz)
        except Exception:
            continue

        n_j = x_j * nH               # cm^-3
        m_j = float(m_j)             # g

        # same gamma-law as your sigma_O
        gamma_j = 1.9e-9 / (m_j + m_n)

        # beta_j = |Z| e B / (m c) * 1/(gamma rho_n)
        beta_j = (abs(Z_j) * e_cgs * Bsafe) / (m_j * c_cgs) * (1.0 / (gamma_j * rho_n))

        # contributions
        denom = 1.0 + beta_j**2
        sumP += n_j * Z_j * (beta_j / denom)
        sumH += n_j * Z_j * (1.0 / denom)

        # for D if desired
        rho_i += n_j * m_j

    # sigma_P, sigma_H
    pref = (e_cgs * c_cgs) / Bsafe
    sigma_P = pref * sumP
    sigma_H = pref * sumH

    # total perpendicular conductivity
    sigma_perp = np.sqrt(sigma_P**2 + sigma_H**2)

    # sigma_O ions-only (your definition)
    sigma_O =  sigma_electron(model, model.nHtot, model.tg)

    # Ohmic resistivity
    eta_O = (c_cgs**2) / (4.0*pi*sigma_O)

    # D factor: rho_neu / rho_gas. If you only have ions+neutrals, rho_gas = rho_n + rho_i.
    # If you want to ignore the correction, set D=1.
    rho_gas = rho_n + rho_i
    D = rho_n / rho_gas

    # Ambipolar resistivity
    # eta_A = D^2 * ( c^2/(4pi) * sigma_P / sigma_perp^2 - eta_O )
    eta_A = (D**2) * ( (c_cgs**2)/(4.0*pi) * (sigma_P / (sigma_perp**2)) - eta_O )

    return {
        "sigma_O": sigma_O,
        "sigma_P": sigma_P,
        "sigma_H": sigma_H,
        "sigma_perp": sigma_perp,
        "eta_O": eta_O,
        "eta_A": eta_A,
        "D": D,
        "rho_n": rho_n,
        "rho_i": rho_i,
    }

import numpy as np

# cgs constants
e_cgs = 4.803204712570263e-10   # statC
c_cgs = 2.99792458e10           # cm/s
pi = np.pi
mH = 1.6735575e-24              # g
m_e = 9.10938356e-28            # g


def Ambipolar_conductivities_electrons(model, nH, Tgas, B, m_n=2.2*mH):
    """
    Electrons-only:
      sigma_P, sigma_H, sigma_perp using beta_e
      sigma_O,e (Ohmic) using Z^2 form
      eta_O,e = c^2/(4pi*sigma_O,e)
      eta_A,e = D^2 * ( c^2/(4pi) * sigma_P / sigma_perp^2 - eta_O,e )

    gamma_e = 8.28e-10 * sqrt(Tgas) / (m_e + m_n)   [as requested]
    """
    # neutral mass density
    rho_n = nH * m_n  # g cm^-3

    # electron abundance and density
    x_e = model.getAbun("e-")      # (nx,nz) n_e / nH
    n_e = x_e * nH                   # cm^-3

    # safety for B=0
    Bsafe = np.where(B > 0, B, np.nan)

    Z_e = 1

    # gamma_e (your requested form)
    gamma_e = 8.28e-10 * np.sqrt(Tgas) / (m_e + m_n)

    # beta_e
    beta_e = (abs(Z_e) * e_cgs * Bsafe) / (m_e * c_cgs) * (1.0 / (gamma_e * rho_n))

    denom = 1.0 + beta_e**2

    # Pedersen & Hall sums reduce to single species term: n_e Z_e * (...)
    sumP = n_e * Z_e * (beta_e / denom)
    sumH = n_e * Z_e * (1.0 / denom)

    pref = (e_cgs * c_cgs) / Bsafe
    sigma_P = pref * sumP
    sigma_H = pref * sumH

    sigma_perp = np.sqrt(sigma_P**2 + sigma_H**2)

    # Ohmic conductivity (electrons-only, Z^2 = 1)
    sigma_Oe = sigma_electron(model, model.nHtot, model.tg)
    # Ohmic resistivity
    eta_Oe = (c_cgs**2) / (4.0*pi*sigma_Oe)

    # D factor: rho_neu / rho_gas ~ 1 for electrons
    rho_e = n_e * m_e
    rho_gas = rho_n + rho_e
    D = rho_n / rho_gas

    # Ambipolar resistivity
    eta_Ae = (D**2) * ((c_cgs**2)/(4.0*pi) * (sigma_P / (sigma_perp**2)) - eta_Oe)

    return {
        "sigma_Oe": sigma_Oe,
        "sigma_Pe": sigma_P,
        "sigma_He": sigma_H,
        "sigma_perp_e": sigma_perp,
        "eta_Oe": eta_Oe,
        "eta_Ae": eta_Ae,
        "beta_e": beta_e,
        "gamma_e": gamma_e,
        "D": D,
        "rho_n": rho_n,
        "rho_e": rho_e,
    }
```
