```python

def Rliste(model):
    Rlist=[]
    for i in range(len(model.x)):
        Rlist.append(model.x[i][0])
    return np.array(Rlist)

def GasPressureMidPlane(model):
    Pmid=np.empty((model.tg[0].size, model.tg[0].size))
    Ptemp=[]
    for i in range (len(model.x)):
        Ptemp.append(model.tg[i][0]*model.nHtot[i][0])
    for j in range (len(model.tg[0])):
        for i in range (len(model.tg)):
            Pmid[i,j]=Ptemp[i] 
    Pmid=np.array(Pmid)
    return Pmid



def BetaPlasma(model):
    Pmid=GasPressureMidPlane(model)
    betamag=10**4*model.nHtot*model.tg/Pmid  # Plasma beta
    return betamag

def SoundSpeed(model):
     soundspeed=np.sqrt(kB/(1.5*1.6*1E-24)*model.tg)  # sound speed in cm/s, assuming mean molecular weight of 1.5 (mostly H2 with some He)
     return soundspeed
def SoundSpeedElectron(model):
     soundspdeedeu=np.sqrt(kB/(me)*model.tg)  # sound speed in cm/s, for electrons
     return soundspdeedeu

def AlfvenSpeed(model):
    betamag=BetaPlasma(model)
    soundspeed=SoundSpeed(model)
    Alfvenspeed=(2/betamag)**0.5*soundspeed
    return Alfvenspeed
def AlfvenSpeedElectron(model):
    betamag=BetaPlasma(model)
    soundspdeedeu=SoundSpeedElectron(model)
    AlfvenspeedElectron=(2/betamag)**0.5*soundspdeedeu
    return AlfvenspeedElectron

def AmbipolarElsasser(model): # Am. # Eq 19 B24
    return model.x**(3/2)*model.getAbun("e-")/1E-8*(model.nHtot/1e10)

def ReynoldsNumber(model): #Rm
    return model.x**(3/2)*model.getAbun("e-")/1E-13*(model.tg/100)**0.5
 
def Lambda_0(model): # Appendix B26
    betamag=BetaPlasma(model)
    
    return model.x**(3/2)*model.getAbun("e-")/1E-9*(model.tg/100)**0.5*(10**4/betamag) #*soundspeed/Alfvenspeed (get abun:ne/nhtot)



def MagneticIntensity(model):
    betamag=BetaPlasma(model)
    return np.sqrt(8*np.pi*model.nHtot*kB*model.tg/betamag)


def IonSoundRadius(model):
    betamag=BetaPlasma(model)
    return np.sqrt(mp*c_light**2/(4*np.pi))/ecgs* betamag**0.5 * model.nHtot**(-0.5)


def AlphaVisc(model):
    betamag=BetaPlasma(model)
    alphai=np.where(betamag>1,(2/betamag)**0.5,10**(-1.9+0.57*(np.arctan((4.2-np.log10(betamag))/0.5))**(-1)))

    MRIactiv=np.where(betamag**0.5*Lambda_0(model)<1,0,1)

    delta=0.5
    minval = np.where(Lambda_0(model) < 1, Lambda_0(model), 1)
    alphai=np.where(betamag>1,(2/betamag)**0.5,10**(-1.9+0.57*(np.arctan((4.2-np.log10(betamag))/0.5))**(-1)))
    
    MRIactiv=np.where(betamag**0.5*Lambda_0(model)<1,0,1)
    return MRIactiv*alphai*minval*((50/AmbipolarElsasser(model)**1.2)**2+(8/AmbipolarElsasser(model)**0.3+1)**2)**(-delta)
    
def AlfvenMachNumber(model):
    betamag=BetaPlasma(model)
    alpha=AlphaVisc(model)
    
    return (2*alpha*betamag)**0.5

def MAccretionRate(model):
    TemperatureGas=model.tg
    densiteHmoyliste=model.nHtot
    alphaviscl=AlphaVisc(model)
    Rl=Rliste(model)
    grilleZAUliste=model.z
    Mstar=model.params["mstar"]
    integlist=alphaviscl*densiteHmoyliste*TemperatureGas
    
    #NcolVert=integrate.trapz((densiteEliste+densiteHatliste+densiteH2liste),grilleZAUliste*autocm)
    #Macc=(1-np.sqrt((Mstar/msol)/Rl))**(-1)*3600*24*365/msol*6*np.pi*kB/(np.sqrt(G*0.7*msol))*(Rl*autocm)**(3/2)*integrate.trapezoid(integlist,grilleZAUliste*autocm,axis=1)
    Macc=3600*24*365/msol*6*np.pi*kB/(np.sqrt(G*Mstar))*(Rl*autocm)**(3/2)*integrate.trapezoid(integlist,grilleZAUliste*autocm,axis=1)
    return Macc
def MAccretionRateSS(model,alpha,L0): # constant alpha viscous model no dead zone
    TemperatureGas=model.tg
    densiteHmoyliste=model.nHtot
    
    Rl=Rliste(model)
    grilleZAUliste=model.z
    Mstar=model.params["mstar"]
    
    integlist=densiteHmoyliste*TemperatureGas*alpha
    #NcolVert=integrate.trapz((densiteEliste+densiteHatliste+densiteH2liste),grilleZAUliste*autocm)
    #Macc=(1-np.sqrt((Mstar/msol)/Rl))**(-1)*3600*24*365/msol*6*np.pi*kB/(np.sqrt(G*0.7*msol))*(Rl*autocm)**(3/2)*integrate.trapezoid(integlist,grilleZAUliste*autocm,axis=1)
    Macc=L0/0.1*3600*24*365/msol*6*np.pi*kB/(np.sqrt(G*Mstar))*(Rl*autocm)**(3/2)*integrate.trapezoid(integlist,grilleZAUliste*autocm,axis=1)
    return Macc


def OmegaKeplerian(model):
     Mstar=model.params["mstar"]
     return 2*1E-7*(Mstar/msol)**(1/2)*model.x**-(3/2) #for a Mstar= 0.7 Msun

def OmegaCyclotron(model):
     MagneticIntensity=MagneticIntensity(model)
     return ecgs*MagneticIntensity/(mp*c_light)

def ScaleHeight(model):
     soundspeed=SoundSpeed(model)
     OmegaKepl=OmegaKeplerian(model)
     return soundspeed/OmegaKepl

def TurbulentVelocity(model):
     soundspeed=SoundSpeed(model)
     return AlphaVisc(model)**0.5*soundspeed  #see B26 for the index of the power law of the dependence of alpha .

def ThermalEnergy(model):
    
    return 3/2*kEneV*model.tg

def LengthScaleTurbulenceEddy(model):
    return np.sqrt(AlphaVisc(model))*ScaleHeight(model)

def AccelerationRate(model):
    return np.full_like(model.x, 2/3) #tacc^-1 in unit ta

def InjectionTimeScale(model):
    return AlfvenMachNumber(model)**-2 #tinj in unit ta
#Magnetic Reconnection parameters
def ReconnectionWidthDelta(model):
    return AlfvenMachNumber(model)**2*LengthScaleTurbulenceEddy(model)

def EscapeTimeScale(model):
    return np.full_like(model.x, 1) #tesc in unit ta

def BetaAcc(model):
    return 1/(AccelerationRate(model)*EscapeTimeScale(model))

#print(betaacc)
def AccelerationDuration(model):
    return AlfvenMachNumber(model)**-1 #Acceleration process duration in unit ta

def EnergyMaximumTurbulentEddyTime(model):
    return np.exp(AccelerationRate(model)*AccelerationDuration(model))

def EnergyMaximumHillas(model):
    return ecgs*ReconnectionWidthDelta(model)*MagneticIntensity(model)*ergtoeV/ThermalEnergy(model)

def EnergyMaximum(model):
    emax=np.minimum(EnergyMaximumHillas(model),EnergyMaximumTurbulentEddyTime(model))
    return emax
    #return np.where(BetaPlasma(model)<10, emax, ThermalEnergy(model)) #Maximum energy in unit Eth


import numpy as np
from scipy.constants import c, epsilon_0, e, m_e

def skin_depth(n, m=m_e):
    
    omega_p = np.sqrt(n * e**2 / (epsilon_0 * m))
    delta = c / omega_p
    
    return delta
```

```python
'''
import numpy as np
import pandas as pd

# -------------------------------------------------------------------
# USER SETTINGS (edit these)
# -------------------------------------------------------------------
SPECIES = ["e-", "HCN"]

# Define the "chemically relevant" layer in terms of z/r (default: your earlier range)
ZR_MIN, ZR_MAX = 0.05, 0.20

# Define the radial range (AU)
R_MIN, R_MAX = 0.10, 3.0
# Threshold (dex) to define "significant EP impact"
DEX_THR = 0.30  # i.e. factor 2

# When comparing parameter effects, we summarise each model with these metrics
# You can add/remove metrics later.
# -------------------------------------------------------------------

def get_r_2d(model):
    """Return R (AU) as a 2D array matching getAbun() shape."""
    return np.asarray(model.x)
# -------------------------------------------------------------------
# Helpers: robust accessors for grids and abundances
# -------------------------------------------------------------------
def _get_attr_first(obj, candidates):
    for name in candidates:
        if hasattr(obj, name):
            return getattr(obj, name)
    return None

def get_z_over_r(model):
    """
    Return z/r as a 2D array matching getAbun() shape.

    In your case the ProDiMo reader stores the cylindrical coordinates as:
      - model.x : 2D array of R (AU)
      - model.z : 2D array of Z (AU)
    """
    R = np.asarray(model.x)
    Z = np.asarray(model.z)

    if R.shape != Z.shape:
        raise ValueError(f"model.x and model.z must have same shape, got {R.shape} vs {Z.shape}")

    return Z / np.maximum(R, 1e-30)

def get_abun(model, species):
    arr = np.asarray(model.getAbun(species))
    return arr

def safe_log10(x, floor=1e-99):
    return np.log10(np.maximum(x, floor))


# -------------------------------------------------------------------
# Metrics for a single model (NOEP or EP)
# -------------------------------------------------------------------
def compute_layer_metrics(model, species, zr_min=ZR_MIN, zr_max=ZR_MAX, r_min=R_MIN, r_max=R_MAX):
    """
    Summary metrics of log10 abundance within a (z/r, R) selection window.
    """
    zr = get_z_over_r(model)
    R  = get_r_2d(model)
    y  = get_abun(model, species)

    if zr.shape != y.shape or R.shape != y.shape:
        raise ValueError(f"Shape mismatch: zr {zr.shape}, R {R.shape}, {species} {y.shape}")

    mask = (
        (zr >= zr_min) & (zr <= zr_max) &
        (R  >= r_min)  & (R  <= r_max) &
        np.isfinite(y)
    )

    if not np.any(mask):
        return dict(
            layer_mean_log10=np.nan,
            layer_max_log10=np.nan,
            layer_p95_log10=np.nan,
            peak_zr=np.nan,
        )

    logy = safe_log10(y)
    vals = logy[mask]

    idx = np.nanargmax(np.where(mask, y, -np.inf))
    peak_zr = float(zr.ravel()[idx])

    return dict(
        layer_mean_log10=float(np.nanmean(vals)),
        layer_max_log10=float(np.nanmax(vals)),
        layer_p95_log10=float(np.nanpercentile(vals, 95)),
        peak_zr=peak_zr,
    )


def compute_ep_effect_metrics(model_noep, model_ep, species,
                              zr_min=ZR_MIN, zr_max=ZR_MAX, r_min=R_MIN, r_max=R_MAX, dex_thr=DEX_THR):
    """
    EP impact metrics within a (z/r, R) selection window:
      drel = log10(y_EP / y_NOEP)

    pen_zr: minimum z/r in the window where drel > dex_thr
    frac_area_sig: fraction of cells in the window where drel > dex_thr
    """
    zr = get_z_over_r(model_noep)
    R  = get_r_2d(model_noep)
    y0 = get_abun(model_noep, species)
    y1 = get_abun(model_ep, species)

    if zr.shape != y0.shape or y0.shape != y1.shape or R.shape != y0.shape:
        raise ValueError(f"Shape mismatch: zr {zr.shape}, R {R.shape}, y0 {y0.shape}, y1 {y1.shape}")

    mask = (
        (zr >= zr_min) & (zr <= zr_max) &
        (R  >= r_min)  & (R  <= r_max) &
        np.isfinite(y0) & np.isfinite(y1)
    )

    if not np.any(mask):
        return dict(
            drel_mean=np.nan,
            drel_max=np.nan,
            drel_p95=np.nan,
            pen_zr=np.nan,
            frac_area_sig=np.nan,
        )

    drel = safe_log10(y1) - safe_log10(y0)
    vals = drel[mask]

    sig = mask & (drel > dex_thr)
    pen_zr = float(np.nanmin(zr[sig])) if np.any(sig) else np.nan
    frac_area_sig = float(np.sum(sig) / np.sum(mask))

    return dict(
        drel_mean=float(np.nanmean(vals)),
        drel_max=float(np.nanmax(vals)),
        drel_p95=float(np.nanpercentile(vals, 95)),
        pen_zr=pen_zr,
        frac_area_sig=frac_area_sig,
    )

# -------------------------------------------------------------------
# Build a tidy dataframe from your model lists
# -------------------------------------------------------------------
def parse_key(model):
    p = model.params
    return (float(p["lacc"]), int(p["settle_method"]), float(p["a_settle"]))

def build_model_maps(modelsNOEP, modelsEP):
    d0 = {parse_key(m): m for m in modelsNOEP}
    d1 = {parse_key(m): m for m in modelsEP}
    missing = sorted(set(d0.keys()) ^ set(d1.keys()))
    if missing:
        raise KeyError(f"NOEP/EP grids do not match. Missing keys: {missing[:10]} ...")
    return d0, d1

def make_metrics_table(modelsNOEP, modelsEP, species_list=SPECIES):
    d0, d1 = build_model_maps(modelsNOEP, modelsEP)

    rows = []
    for (LA, sm, a) in sorted(d0.keys()):
        m0 = d0[(LA, sm, a)]
        m1 = d1[(LA, sm, a)]
        base = dict(Lacc=LA, setmet=sm, a_settle=a, name_noep=getattr(m0, "name", ""), name_ep=getattr(m1, "name", ""))

        row = dict(base)
        for sp in species_list:
            m0m = compute_layer_metrics(m0, sp)
            m1m = compute_layer_metrics(m1, sp)
            dem = compute_ep_effect_metrics(m0, m1, sp)

            # prefix columns by species
            for k, v in m0m.items():
                row[f"{sp}_NOEP_{k}"] = v
            for k, v in m1m.items():
                row[f"{sp}_EP_{k}"] = v
            for k, v in dem.items():
                row[f"{sp}_EPimpact_{k}"] = v

        rows.append(row)

    return pd.DataFrame(rows)


# -------------------------------------------------------------------
# Factorial main effects + interactions via paired differencing
# -------------------------------------------------------------------
def paired_effect(df, metric_col, factor):
    """
    Compute average main effect of 'factor' on metric_col via within-cell pairing.

    factor in {"Lacc", "setmet", "a_settle"}.
    For Lacc (3 levels) returns pairwise effects vs a reference (lowest level).
    For binary factors returns high-low.
    """
    out = []

    if factor == "Lacc":
        levels = sorted(df["Lacc"].unique())
        ref = levels[0]
        for lvl in levels[1:]:
            # pair within (setmet, a_settle)
            merged = df[df["Lacc"] == ref].merge(
                df[df["Lacc"] == lvl],
                on=["setmet", "a_settle"],
                suffixes=("_ref", "_lvl"),
                how="inner",
            )
            delta = merged[f"{metric_col}_lvl"] - merged[f"{metric_col}_ref"]
            out.append(dict(factor=f"{factor}:{lvl}-vs-{ref}", effect=float(np.nanmean(delta)), n=int(delta.notna().sum())))
        return pd.DataFrame(out)

    elif factor in ["setmet", "a_settle"]:
        levels = sorted(df[factor].unique())
        if len(levels) != 2:
            raise ValueError(f"{factor} expected to have 2 levels, got {levels}")
        low, high = levels[0], levels[1]

        # pair within the other factors
        other = ["Lacc", "a_settle"] if factor == "setmet" else ["Lacc", "setmet"]
        merged = df[df[factor] == low].merge(
            df[df[factor] == high],
            on=other,
            suffixes=("_low", "_high"),
            how="inner",
        )
        delta = merged[f"{metric_col}_high"] - merged[f"{metric_col}_low"]
        return pd.DataFrame([dict(factor=f"{factor}:{high}-{low}", effect=float(np.nanmean(delta)), n=int(delta.notna().sum()))])

    else:
        raise ValueError("Unknown factor: " + factor)

def interaction_effect_binary(df, metric_col, factorA, factorB):
    """
    Interaction term for two binary factors (e.g. a_settle x setmet) on metric_col.

    I = (highA-lowA at highB) - (highA-lowA at lowB), averaged over the remaining factor (Lacc).
    """
    levelsA = sorted(df[factorA].unique())
    levelsB = sorted(df[factorB].unique())
    if len(levelsA) != 2 or len(levelsB) != 2:
        raise ValueError("interaction_effect_binary expects both factors to be binary.")
    A0, A1 = levelsA[0], levelsA[1]
    B0, B1 = levelsB[0], levelsB[1]

    # Remaining factor(s)
    remaining = [c for c in ["Lacc", "setmet", "a_settle"] if c not in [factorA, factorB]]

    # Compute deltaA at each B level, then difference
    def deltaA_at_B(Bval):
        sub = df[df[factorB] == Bval]
        merged = sub[sub[factorA] == A0].merge(
            sub[sub[factorA] == A1],
            on=remaining,
            suffixes=("_A0", "_A1"),
            how="inner",
        )
        return merged[f"{metric_col}_A1"] - merged[f"{metric_col}_A0"]

    dB1 = deltaA_at_B(B1)
    dB0 = deltaA_at_B(B0)

    # Need matching remaining keys to subtract properly:
    # do it by merging on remaining
    tmp1 = df[(df[factorA] == A0) & (df[factorB] == B1)][remaining + [metric_col]].rename(columns={metric_col: "M00"})
    tmp2 = df[(df[factorA] == A1) & (df[factorB] == B1)][remaining + [metric_col]].rename(columns={metric_col: "M10"})
    tmp3 = df[(df[factorA] == A0) & (df[factorB] == B0)][remaining + [metric_col]].rename(columns={metric_col: "M01"})
    tmp4 = df[(df[factorA] == A1) & (df[factorB] == B0)][remaining + [metric_col]].rename(columns={metric_col: "M11"})

    merged = tmp1.merge(tmp2, on=remaining).merge(tmp3, on=remaining).merge(tmp4, on=remaining)
    # For B1: (A1-A0) = M10-M00 ; For B0: (A1-A0)=M11-M01
    I = (merged["M10"] - merged["M00"]) - (merged["M11"] - merged["M01"])

    return pd.DataFrame([dict(interaction=f"{factorA}x{factorB}", effect=float(np.nanmean(I)), n=int(I.notna().sum()))])

'''
'''
# -------------------------------------------------------------------
# RUN: build tables + scoreboards
# -------------------------------------------------------------------
df = make_metrics_table(modelsNOEP, modelsEP, species_list=SPECIES)

# Choose which metrics define "most affecting"
# Baseline sensitivity (NOEP): use mean log10 abundance in layer
baseline_metrics = {
    "e-": "e-_NOEP_layer_mean_log10",
    "HCN": "HCN_NOEP_layer_mean_log10",
}

# EP leverage: use mean delta_rel in layer (dex)
epimpact_metrics = {
    "e-": "e-_EPimpact_drel_mean",
    "HCN": "HCN_EPimpact_drel_mean",
}

# --- Scoreboard A: baseline controls (NOEP) ---
score_baseline = []
for sp, col in baseline_metrics.items():
    for fac in ["Lacc", "setmet", "a_settle"]:
        eff = paired_effect(df, col, fac)
        eff.insert(0, "species", sp)
        eff.insert(1, "metric", col)
        score_baseline.append(eff)
score_baseline = pd.concat(score_baseline, ignore_index=True)

# --- Scoreboard B: EP leverage ---
score_ep = []
for sp, col in epimpact_metrics.items():
    for fac in ["Lacc", "setmet", "a_settle"]:
        eff = paired_effect(df, col, fac)
        eff.insert(0, "species", sp)
        eff.insert(1, "metric", col)
        score_ep.append(eff)
score_ep = pd.concat(score_ep, ignore_index=True)

# --- Key interaction: a_settle x setmet on EP impact (binary x binary) ---
interactions = []
for sp, col in epimpact_metrics.items():
    inter = interaction_effect_binary(df, col, "a_settle", "setmet")
    inter.insert(0, "species", sp)
    inter.insert(1, "metric", col)
    interactions.append(inter)
interactions = pd.concat(interactions, ignore_index=True)

# Sort for readability: largest absolute effect first
score_baseline["abs_effect"] = score_baseline["effect"].abs()
score_ep["abs_effect"] = score_ep["effect"].abs()
interactions["abs_effect"] = interactions["effect"].abs()

score_baseline_sorted = score_baseline.sort_values(["species", "abs_effect"], ascending=[True, False])
score_ep_sorted = score_ep.sort_values(["species", "abs_effect"], ascending=[True, False])
interactions_sorted = interactions.sort_values(["species", "abs_effect"], ascending=[True, False])

# -------------------------------------------------------------------
# Output: print compact summaries
# -------------------------------------------------------------------
pd.set_option("display.max_rows", 200)
pd.set_option("display.max_columns", 200)
pd.set_option("display.width", 160)

print("\n======================")
print("MODEL METRICS (first rows)")
print("======================")
print(df[["Lacc","setmet","a_settle",
          "e-_NOEP_layer_mean_log10","HCN_NOEP_layer_mean_log10",
          "e-_EPimpact_drel_mean","HCN_EPimpact_drel_mean",
          "e-_EPimpact_pen_zr","HCN_EPimpact_pen_zr",
          "e-_EPimpact_frac_area_sig","HCN_EPimpact_frac_area_sig"
         ]].head(12))

print("\n======================")
print("SCOREBOARD A: Baseline control (NOEP) — mean log10 in layer")
print("======================")
print(score_baseline_sorted[["species","factor","effect","n","metric"]])

print("\n======================")
print("SCOREBOARD B: EP leverage — mean log10(EP/NOEP) in layer (dex)")
print("======================")
print(score_ep_sorted[["species","factor","effect","n","metric"]])

print("\n======================")
print("INTERACTION: a_settle × setmet on EP leverage")
print("======================")
print(interactions_sorted[["species","interaction","effect","n","metric"]])

# Optional: identify "best EP-impact conditions" per species (largest mean enhancement)
for sp in ["e-","HCN"]:
    col = epimpact_metrics[sp]
    idx = df[col].idxmax()
    row = df.loc[idx, ["Lacc","setmet","a_settle", col, f"{sp}_EPimpact_pen_zr", f"{sp}_EPimpact_frac_area_sig"]]
    print(f"\n>>> Strongest EP impact for {sp} (by {col}):")
    print(row.to_string())

'''
```

```python
import os
import numpy as np
import pandas as pd
import contextlib, io

# =========================
# USER PARAMETERS
# =========================
WL_MIN, WL_MAX = 5.0, 25.0      # micron
FLUX_MIN = 1e-21              # W/m^2 (line included if >= this)
REL_TOL = 0.0                  # for optional filtering on relative differences later

# =========================
# Quiet + robust selector
# =========================
def select_lines_safe(model_obj, ident, wlmin, wlmax):
    try:
        try:
            L = model_obj.selectLineEstimates(ident=ident, wlrange=[wlmin, wlmax])
        except TypeError:
            L = model_obj.selectLineEstimates(ident=ident)
            # force evaluation (some objects lazily compute)
            _ = getattr(L, "flux_Jy", None)
            L = [x for x in L if wlmin <= float(x.wl) <= wlmax]
        return L if L is not None else []
    except Exception:
        return []

def select_lines_quiet(model_obj, ident, wlmin, wlmax):
    f = io.StringIO()
    with contextlib.redirect_stdout(f), contextlib.redirect_stderr(f):
        return select_lines_safe(model_obj, ident, wlmin, wlmax)

# =========================
# Read idents from FlineEstimates.out
# =========================
def idents_from_FlineEstimates(path):
    idents = set()
    with open(path, "r") as f:
        for line in f:
            s = line.strip()
            if (not s) or s.startswith("#"):
                continue
            tok = s.split()
            if tok:
                idents.add(tok[0].strip())
    return sorted(idents)

def get_model_dir(m):
    return m.model_dir

def get_idents_for_pair(model_noep, model_ep):
    # Use the directory stored on the model object
    d0 = get_model_dir(model_noep)
    d1 = get_model_dir(model_ep)

    p0 = os.path.join(d0, "FlineEstimates.out")
    p1 = os.path.join(d1, "FlineEstimates.out")

    id0 = set(idents_from_FlineEstimates(p0))
    id1 = set(idents_from_FlineEstimates(p1))
    return sorted(id0.union(id1))
# =========================
# Flux aggregation
# =========================
def sum_line_flux_per_species(model_obj, idents, wlmin=WL_MIN, wlmax=WL_MAX, flux_min=FLUX_MIN):
    """
    Sum total line flux per species (species = ident split at first underscore).
    Returns dict {species: total_flux}.
    """
    species_flux = {}
    for ident in idents:
        species = ident.split("_")[0]

        lines = select_lines_quiet(model_obj, ident, wlmin, wlmax)
        if not lines:
            continue

        tot = 0.0
        for l in lines:
            F = float(l.flux)
            if F >= flux_min:
                tot += F

        if tot > 0.0:
            species_flux[species] = species_flux.get(species, 0.0) + tot

    return species_flux

def sum_total_flux(model_obj, idents, wlmin=WL_MIN, wlmax=WL_MAX, flux_min=FLUX_MIN):
    """
    Sum total flux over all idents/lines in range. Returns float.
    """
    tot = 0.0
    for ident in idents:
        lines = select_lines_quiet(model_obj, ident, wlmin, wlmax)
        for l in lines:
            F = float(l.flux)
            if F >= flux_min:
                tot += F
    return tot

def rel_diff(F_ep, F_noep):
    if F_noep <= 0:
        return np.nan
    return (F_ep - F_noep) / F_noep

# =========================
# Main: compare a LIST of matched models (12 combos)
# =========================
def compare_line_fluxes(modelsNOEP, modelsEP, wlmin=WL_MIN, wlmax=WL_MAX, flux_min=FLUX_MIN):
    """
    For each parameter-combination, compute:
      - total flux (all species) NOEP and EP
      - per-species summed fluxes and relative differences

    Returns
    -------
    df_total : pandas.DataFrame
      one row per model with global totals
    df_species : pandas.DataFrame
      long-format table: one row per (model, species)
    """
    # Map by parameter key (same as your chemistry script)
    def parse_key(model):
        p = model.params
        return (float(p["lacc"]), int(p["settle_method"]), float(p["a_settle"]))

    d0 = {parse_key(m): m for m in modelsNOEP}
    d1 = {parse_key(m): m for m in modelsEP}
    keys = sorted(set(d0.keys()) & set(d1.keys()))
    missing = sorted(set(d0.keys()) ^ set(d1.keys()))
    if missing:
        raise KeyError(f"NOEP/EP grids do not match. Missing keys: {missing[:10]} ...")

    total_rows = []
    species_rows = []

    for (LA, sm, a) in keys:
        m0 = d0[(LA, sm, a)]
        m1 = d1[(LA, sm, a)]

        # Use union of idents from BOTH models
        idents = get_idents_for_pair(m0, m1)

        # Totals (all lines)
        F0_tot = sum_total_flux(m0, idents, wlmin=wlmin, wlmax=wlmax, flux_min=flux_min)
        F1_tot = sum_total_flux(m1, idents, wlmin=wlmin, wlmax=wlmax, flux_min=flux_min)

        total_rows.append(dict(
            Lacc=LA, setmet=sm, a_settle=a,
            totalFlux_NOEP=F0_tot,
            totalFlux_EP=F1_tot,
            totalFlux_rel=rel_diff(F1_tot, F0_tot),
            name_noep=getattr(m0, "name", ""),
            name_ep=getattr(m1, "name", ""),
        ))

        # Per-species
        sp0 = sum_line_flux_per_species(m0, idents, wlmin=wlmin, wlmax=wlmax, flux_min=flux_min)
        sp1 = sum_line_flux_per_species(m1, idents, wlmin=wlmin, wlmax=wlmax, flux_min=flux_min)

        all_species = sorted(set(sp0.keys()) | set(sp1.keys()))
        for sp in all_species:
            F0 = sp0.get(sp, 0.0)
            F1 = sp1.get(sp, 0.0)
            if (F0 < flux_min) and (F1 < flux_min):
                continue

            species_rows.append(dict(
                Lacc=LA, setmet=sm, a_settle=a,
                species=sp,
                flux_NOEP=F0,
                flux_EP=F1,
                rel=rel_diff(F1, F0),
            ))

    df_total = pd.DataFrame(total_rows).sort_values(["Lacc", "setmet", "a_settle"]).reset_index(drop=True)
    df_species = pd.DataFrame(species_rows).sort_values(["Lacc", "setmet", "a_settle", "species"]).reset_index(drop=True)

    return df_total, df_species


# =========================
# RUN (assuming modelsNOEP and modelsEP already exist)
# =========================
df_total, df_species = compare_line_fluxes(modelsNOEP, modelsEP, wlmin=WL_MIN, wlmax=WL_MAX, flux_min=FLUX_MIN)

print("\n======================")
print("TOTAL line flux (all species, summed) per model")
print("======================")
print(df_total[["Lacc","setmet","a_settle","totalFlux_NOEP","totalFlux_EP","totalFlux_rel"]])

print("\n======================")
print("Top EP/NOEP relative differences by species (global across all 12)")
print("======================")
tmp = df_species.replace([np.inf, -np.inf], np.nan).dropna(subset=["rel"]).copy()
tmp["abs_rel"] = tmp["rel"].abs()
print(tmp.sort_values("abs_rel", ascending=False).head(30)[["Lacc","setmet","a_settle","species","flux_NOEP","flux_EP","rel"]])

# Optional: if you only want large relative changes
if REL_TOL > 0:
    tmp2 = tmp[tmp["abs_rel"] >= REL_TOL]
    print(f"\nSpecies rows with |rel| >= {REL_TOL}: {len(tmp2)}")
```

```python
import os

def get_idents_for_pair(model1, model2):
    d0 = get_model_dir(model1)
    d1 = get_model_dir(model2)

    p0 = os.path.join(d0, "FlineEstimates.out")
    p1 = os.path.join(d1, "FlineEstimates.out")

    id0 = set(idents_from_FlineEstimates(p0))
    id1 = set(idents_from_FlineEstimates(p1))

    return sorted(id0.union(id1))


def compute_species_fluxes_for_pair(model_noep, model_ep, wlmin, wlmax, flux_min=1e-21):
    """
    Compute integrated flux per species for NOEP and EP using the same ident list.
    """
    idents = get_idents_for_pair(model_noep, model_ep)

    flux_noep = sum_line_flux_per_species(
        model_noep, idents,
        wlmin=wlmin, wlmax=wlmax, flux_min=flux_min
    )

    flux_ep = sum_line_flux_per_species(
        model_ep, idents,
        wlmin=wlmin, wlmax=wlmax, flux_min=flux_min
    )

    return flux_noep, flux_ep


def compute_all_ratios(flux_dict):
    """
    Compute all possible ordered ratios species1/species2 from a flux dictionary.
    Returns a dict {(species1, species2): ratio}.
    """
    ratios = {}
    species_list = sorted(flux_dict.keys())

    for sp1 in species_list:
        for sp2 in species_list:
            if sp1 == sp2:
                continue

            f1 = flux_dict.get(sp1, 0.0)
            f2 = flux_dict.get(sp2, 0.0)

            if f2 > 0.0:
                ratios[(sp1, sp2)] = f1 / f2
            else:
                ratios[(sp1, sp2)] = None

    return ratios


def compare_all_ratios(model_noep, model_ep, wlmin, wlmax, flux_min=1e-21):
    """
    Compare all possible ordered ratios between NOEP and EP.
    """
    flux_noep, flux_ep = compute_species_fluxes_for_pair(
        model_noep, model_ep,
        wlmin=wlmin, wlmax=wlmax, flux_min=flux_min
    )

    # species présentes dans au moins un des deux modèles
    all_species = sorted(set(flux_noep.keys()).union(set(flux_ep.keys())))

    # on complète avec 0.0 pour éviter les espèces manquantes
    flux_noep_full = {sp: flux_noep.get(sp, 0.0) for sp in all_species}
    flux_ep_full   = {sp: flux_ep.get(sp,   0.0) for sp in all_species}

    ratios_noep = compute_all_ratios(flux_noep_full)
    ratios_ep   = compute_all_ratios(flux_ep_full)

    results = []

    for sp1 in all_species:
        for sp2 in all_species:
            if sp1 == sp2:
                continue

            r_noep = ratios_noep[(sp1, sp2)]
            r_ep   = ratios_ep[(sp1, sp2)]

            if r_noep is None or r_ep is None or r_noep == 0.0:
                rel_change = None
            else:
                rel_change = (r_ep - r_noep) / r_noep

            results.append((sp1, sp2, r_noep, r_ep, rel_change))

    return results, flux_noep_full, flux_ep_full


```

```python
def compare_unique_ratios(model_noep, model_ep, wlmin, wlmax,
                          flux_min=1e-21,
                          exclude=None,
                          priority=("HCN", "C2H2","H","NH3")):

    if exclude is None:
        exclude = []

    flux_noep, flux_ep = compute_species_fluxes_for_pair(
        model_noep, model_ep,
        wlmin=wlmin, wlmax=wlmax, flux_min=flux_min
    )

    # --- Filter species ---
    all_species = sorted(
        set(flux_noep.keys()).union(set(flux_ep.keys()))
    )
    all_species = [sp for sp in all_species if sp not in exclude]

    results = []

    def add_ratio(sp1, sp2):
        f1_noep = flux_noep.get(sp1, 0.0)
        f2_noep = flux_noep.get(sp2, 0.0)
        f1_ep   = flux_ep.get(sp1, 0.0)
        f2_ep   = flux_ep.get(sp2, 0.0)

        r_noep = None if f2_noep == 0.0 else f1_noep / f2_noep
        r_ep   = None if f2_ep   == 0.0 else f1_ep   / f2_ep

        if r_noep is None or r_ep is None or r_noep == 0.0:
            rel_change = None
        else:
            rel_change = (r_ep - r_noep) / r_noep

        results.append((sp1, sp2, r_noep, r_ep, rel_change))

    # --- Loop over UNIQUE pairs only ---
    for i, spA in enumerate(all_species):
        for spB in all_species[i+1:]:

            # --- Decide numerator based on priority ---
            if spA in priority and spB in priority:
                # if both are priority → keep defined order (HCN first)
                sp1, sp2 = (spA, spB) if priority.index(spA) < priority.index(spB) else (spB, spA)

            elif spA in priority:
                sp1, sp2 = spA, spB

            elif spB in priority:
                sp1, sp2 = spB, spA

            else:
                sp1, sp2 = spA, spB

            add_ratio(sp1, sp2)

    return results
```

```python
def plot_ratio_two_models(model_ref,
                          model_new,
                          field_func,
                          label,
                          xlim=(0.07, 100.0),
                          ylim=(0.0, 0.4),
                          zlim=(1e-2, 1e2),
                          av_levels=(1.0,),
                          nh_levels=(5e22,5e23, 5e24),
                          chi_levels=(1.0,),
                          figsize=(6,3),
                          outfile=None):
    """
    Plot ratio between two ProDiMo models:

        ratio = field(model_new) / field(model_ref)

    Parameters
    ----------
    model_ref : reference model (denominator)
    model_new : new model (numerator)
    field_func : callable
        Example: lambda m: m.getAbun("e-")
    label : str
        Label for colorbar
    """

    fig, ax = plt.subplots(figsize=figsize)

    R  = np.asarray(model_ref.x)
    zr = np.asarray(model_ref.z) / np.maximum(R, 1e-30)

    y_ref = np.asarray(field_func(model_ref))
    y_new = np.asarray(field_func(model_new))

    ratio = np.maximum(y_new, 1e-99) / np.maximum(y_ref, 1e-99)

    # mask outside region
    mask = (R >= xlim[0]) & (R <= xlim[1]) & (zr >= ylim[0]) & (zr <= ylim[1])
    ratio = ratio.astype(float, copy=True)
    ratio[~mask] = np.nan

    norm = LogNorm(vmin=zlim[0], vmax=zlim[1])
    pcm = ax.pcolormesh(R, zr, ratio, shading="auto", norm=norm)

    # optional contours from reference model
    if hasattr(model_ref, "AV"):
        ax.contour(R, zr, np.asarray(model_ref.AV),
                   levels=list(av_levels), colors="k", linewidths=0.8)

    if hasattr(model_ref, "NHver"):
        ax.contour(R, zr, np.asarray(model_ref.NHver),
                   levels=list(nh_levels), colors="red", linewidths=0.7)

    if hasattr(model_ref, "chiRT"):
        ax.contour(R, zr, np.asarray(model_ref.chiRT),
                   levels=list(chi_levels), colors="blue", linewidths=0.7)

    ax.set_xscale("log")
    ax.set_xlim(*xlim)
    ax.set_ylim(*ylim)
    ax.set_xlabel(r"$R\,[\mathrm{AU}]$")
    ax.set_ylabel(r"$z/r$")
    #plt.title(rf"1: {model_new.name}  vs  2: {model_ref.name}", fontsize=10)
    cbar = fig.colorbar(pcm)
    #cbar.set_label(rf"${label}^{{1}}/{label}^{{0}}$", fontsize=18)
    cbar.set_label(label, fontsize=18)

    if outfile:
        fig.savefig('Figure/'+outfile, dpi=250, bbox_inches="tight")

    plt.show()
    return None
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm

def plot_ratio_four_models(model_ref,
                           model_new,
                           field_funcs,
                           labels,
                           zlims,
                           titles=None,
                           xlim=(0.07, 100.0),
                           ylim=(0.0, 0.4),
                           av_levels=(1.0,),
                           nh_levels=(5e22, 5e23, 5e24),
                           chi_levels=(1.0,),
                           figsize=(12, 8),
                           outfile=None):
    """
    Plot ratios between two ProDiMo models for four quantities in a 2x2 panel figure.

    Parameters
    ----------
    model_ref : reference model (denominator)
    model_new : new model (numerator)
    field_funcs : list of callables
        Functions like lambda m: m.getAbun("e-")
    labels : list of str
        Colorbar labels
    zlims : list of tuple
        (vmin, vmax) for each panel
    titles : list of str, optional
        Titles for each panel
    """

    fig, axes = plt.subplots(2, 2, figsize=figsize, sharex=True, sharey=True)
    axes = axes.flatten()

    R  = np.asarray(model_ref.x)
    zr = np.asarray(model_ref.z) / np.maximum(R, 1e-30)

    mask = (R >= xlim[0]) & (R <= xlim[1]) & (zr >= ylim[0]) & (zr <= ylim[1])

    for i, ax in enumerate(axes):
        y_ref = np.asarray(field_funcs[i](model_ref))
        y_new = np.asarray(field_funcs[i](model_new))

        ratio = np.maximum(y_new, 1e-99) / np.maximum(y_ref, 1e-99)
        ratio = ratio.astype(float, copy=True)
        ratio[~mask] = np.nan

        if i in [0, 1]:

            norm = None

            vmin, vmax = zlims[i]

        else:

            norm = LogNorm(vmin=zlims[i][0], vmax=zlims[i][1])

            vmin, vmax = None, None
        pcm = ax.pcolormesh(

        R,

        zr,

        ratio,

        shading="auto",

        norm=norm,

        vmin=vmin,

        vmax=vmax

    )

        # optional contours from reference model
        if hasattr(model_ref, "AV"):
            ax.contour(R, zr, np.asarray(model_ref.AV),
                       levels=list(av_levels), colors="k", linewidths=0.8)

        if hasattr(model_ref, "NHver"):
            ax.contour(R, zr, np.asarray(model_ref.NHver),
                       levels=list(nh_levels), colors="red", linewidths=0.7)

        if hasattr(model_ref, "chiRT"):
            ax.contour(R, zr, np.asarray(model_ref.chiRT),
                       levels=list(chi_levels), colors="blue", linewidths=0.7)

        ax.set_xscale("log")
        ax.set_xlim(*xlim)
        ax.set_ylim(*ylim)

        if titles is not None:
            ax.set_title(titles[i], fontsize=12)

        cbar = fig.colorbar(pcm, ax=ax, pad=0.02)
        cbar.set_label(labels[i], fontsize=12)

    # axis labels only on outer panels
    axes[2].set_xlabel(r"$R\,[\mathrm{AU}]$")
    axes[3].set_xlabel(r"$R\,[\mathrm{AU}]$")
    axes[0].set_ylabel(r"$z/r$")
    axes[2].set_ylabel(r"$z/r$")

    plt.tight_layout()

    if outfile:
        fig.savefig("Figure/" + outfile, dpi=250, bbox_inches="tight")

    plt.show()


plot_ratio_four_models(
    model_ref=modelsNOEP[2],
    model_new=modelsEP[2],
    field_funcs=[
        lambda m: m.zetaSTCR + m.zetaX + m.zetaCR,
        lambda m: m.getAbun("e-"),
        lambda m: m.getAbun("HCN"),
        lambda m: m.getAbun("NH3"),
    ],
    labels=[
        r"$\zeta^{\rm EP}/\zeta^{\rm ref}$",
        r"$x_{e^-}^{\rm EP}/x_{e^-}^{\rm ref}$",
        r"$x_{\rm HCN}^{\rm EP}/x_{\rm HCN}^{\rm ref}$",
        r"$x_{\rm NH_3}^{\rm EP}/x_{\rm NH_3}^{\rm ref}$",
    ],
    zlims=[
        (1e0, 10),
        (1e0, 10),
        (1e-1, 1e2),
        (1e-1, 10),
    ],
    titles=[
        r"Ionisation rate",
        r"Electron abundance",
        r"HCN abundance",
        r"NH$_3$ abundance",
    ],
    outfile=None
)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm


def plot_ratio_four_models(model_ref,
                           model_new,
                           field_funcs,
                           labels,
                           zlims,
                           titles=None,
                           xlim=(0.07, 100.0),
                           ylim=(0.0, 0.4),
                           av_levels=(1.0,),
                           nh_levels=(5e22, 5e23, 5e24),
                           chi_levels=(1.0,),
                           line_origins=None,
                           figsize=(15, 10),
                           outfile=None,
                           title_fs=24,
                           label_fs=24,
                           tick_fs=20,
                           cbar_label_fs=22,
                           cbar_tick_fs=18,
                           contour_lw=2):
    """
    Plot ratios between two ProDiMo models for four quantities.

    The top-left panel is produced with pplot.plot_line_origin().
    The other three panels are produced with matplotlib pcolormesh.

    ratio = field(model_new) / field(model_ref)
    """

    if line_origins is None:
        line_origins = [
            ["o-H2O", 21.5],
            ["C2H2_H", 13.5],
            ["HCN_H", 14.0],
            ["NH3_H", 20.0],
        ]

    R = np.asarray(model_ref.x)
    zr = np.asarray(model_ref.z) / np.maximum(R, 1e-30)

    mask = (
        (R >= xlim[0]) &
        (R <= xlim[1]) &
        (zr >= ylim[0]) &
        (zr <= ylim[1])
    )

    # ============================================================
    # Top-left panel: use prodimopy plot_line_origin directly
    # ============================================================

    y_ref_0 = np.asarray(field_funcs[0](model_ref))
    y_new_0 = np.asarray(field_funcs[0](model_new))

    ratio_0 = np.maximum(y_new_0, 1e-99) / np.maximum(y_ref_0, 1e-99)
    ratio_0 = ratio_0.astype(float, copy=True)
    ratio_0[~mask] = np.nan

    pp = pplot.Plot(None)

    cont = pplot.Contour(
        model_ref.NHver,
        list(nh_levels),
        label_fmt=r"N$_H$ = %.0e cm$^{-2}$",
        colors="red",
        linestyles="-",
        linewidths=contour_lw,
        showlabels=True
    )

    cont2 = pplot.Contour(
        model_ref.chiRT,
        list(chi_levels),
        label_fmt=r"$\chi=1$",
        colors="blue",
        linestyles="-",
        linewidths=contour_lw,
        showlabels=True
    )

    cont3 = pplot.Contour(
        model_ref.AV,
        list(av_levels),
        label_fmt=r"A$_V=1$",
        colors="black",
        linestyles="-",
        linewidths=contour_lw,
        showlabels=True
    )

    fig=pp.plot_line_origin(modelsNOEP[2],[['o-H2O',21.52],["C2H2_H",13.5],["HCN_H",14.0],['NH3_H',20.0],],field=modelsEP[2].zetaSTCR/(modelsNOEP[2].zetaSTCR+modelsNOEP[2].zetaX+modelsNOEP[2].zetaCR),
                        label=r"$\zeta_{EP}/\zeta_{X,GCR}$",zlim=[0.5,15],extend="both",cb_format="%.0f",zr=True,xlog=True,zlog=False,
                        xlim=[None,150],ylim=[None,0.5],
                        showContOrigin=True,
                        showRadialLines=False,
                        showztauD1=False,
                        oconts=[cont, cont2, cont3]) # this adds additional contours


    fig.set_size_inches(figsize)

    # Existing axes from plot_line_origin
    ax0 = fig.axes[0]
    cax0 = fig.axes[-1]

    ax0.set_position([0.08, 0.56, 0.34, 0.35])
    cax0.set_position([0.425, 0.56, 0.018, 0.35])

    ax0.tick_params(axis="both", which="major", labelsize=tick_fs, width=1.5, length=7)
    ax0.tick_params(axis="both", which="minor", labelsize=tick_fs, width=1.2, length=4)
    cax0.tick_params(labelsize=cbar_tick_fs, width=1.3, length=6)

    ax0.set_xlabel("")
    ax0.set_ylabel(r"$z/r$", fontsize=label_fs)

    if titles is not None:
        ax0.set_title(titles[0], fontsize=title_fs, pad=12)

    cax0.set_ylabel(labels[0], fontsize=cbar_label_fs, labelpad=12)

    # ============================================================
    # Other three panels
    # ============================================================

    ax1 = fig.add_axes([0.56, 0.56, 0.34, 0.35], sharex=ax0, sharey=ax0)
    ax2 = fig.add_axes([0.08, 0.10, 0.34, 0.35], sharex=ax0, sharey=ax0)
    ax3 = fig.add_axes([0.56, 0.10, 0.34, 0.35], sharex=ax0, sharey=ax0)

    axes = [ax0, ax1, ax2, ax3]

    cax1 = fig.add_axes([0.905, 0.56, 0.018, 0.35])
    cax2 = fig.add_axes([0.425, 0.10, 0.018, 0.35])
    cax3 = fig.add_axes([0.905, 0.10, 0.018, 0.35])

    caxes = [cax0, cax1, cax2, cax3]

    for i, ax in zip([1, 2, 3], [ax1, ax2, ax3]):

        y_ref = np.asarray(field_funcs[i](model_ref))
        y_new = np.asarray(field_funcs[i](model_new))

        ratio = np.maximum(y_new, 1e-99) / np.maximum(y_ref, 1e-99)
        ratio = ratio.astype(float, copy=True)
        ratio[~mask] = np.nan

        norm = LogNorm(vmin=zlims[i][0], vmax=zlims[i][1])

        pcm = ax.pcolormesh(
            R,
            zr,
            ratio,
            shading="auto",
            norm=norm
        )

        if hasattr(model_ref, "AV"):
            ax.contour(
                R,
                zr,
                np.asarray(model_ref.AV),
                levels=list(av_levels),
                colors="black",
                linewidths=contour_lw
            )

        if hasattr(model_ref, "NHver"):
            ax.contour(
                R,
                zr,
                np.asarray(model_ref.NHver),
                levels=list(nh_levels),
                colors="red",
                linewidths=contour_lw
            )

        if hasattr(model_ref, "chiRT"):
            ax.contour(
                R,
                zr,
                np.asarray(model_ref.chiRT),
                levels=list(chi_levels),
                colors="blue",
                linewidths=contour_lw
            )

        ax.set_xscale("log")
        ax.set_xlim(*xlim)
        ax.set_ylim(*ylim)

        ax.tick_params(
            axis="both",
            which="major",
            labelsize=tick_fs,
            width=1.5,
            length=7
        )

        ax.tick_params(
            axis="both",
            which="minor",
            labelsize=tick_fs,
            width=1.2,
            length=4
        )

        if titles is not None:
            ax.set_title(titles[i], fontsize=title_fs, pad=12)

        cbar = fig.colorbar(
            pcm,
            cax=caxes[i]
        )

        cbar.set_label(
            labels[i],
            fontsize=cbar_label_fs,
            labelpad=12
        )

        cbar.ax.tick_params(
            labelsize=cbar_tick_fs,
            width=1.3,
            length=6
        )

    # Axis labels
    ax0.set_ylabel(r"$z/r$", fontsize=label_fs)
    ax2.set_ylabel(r"$z/r$", fontsize=label_fs)

    ax2.set_xlabel(r"$R\,[\mathrm{AU}]$", fontsize=label_fs)
    ax3.set_xlabel(r"$R\,[\mathrm{AU}]$", fontsize=label_fs)

    ax1.set_ylabel("")
    ax3.set_ylabel("")
    ax0.set_xlabel("")
    ax1.set_xlabel("")

    # Hide duplicated tick labels
    plt.setp(ax0.get_xticklabels(), visible=False)
    plt.setp(ax1.get_xticklabels(), visible=False)
    plt.setp(ax1.get_yticklabels(), visible=False)
    plt.setp(ax3.get_yticklabels(), visible=False)

    if outfile:
        fig.savefig(
            "Figure/" + outfile,
            dpi=300,
            bbox_inches="tight"
        )

    plt.show()
```
